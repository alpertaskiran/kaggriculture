## 4. Detecting a phase advance without seeing the current action

Simultaneous actions prevent us from reading the opponent's current
SELL. The next observation does reveal public market inventory.
For item `i`, v45 computes a conservative residual:

```text
unexplained_supply[i]
  = market_inventory_delta[i]
  - my_previous_signed_market_flow[i]
  - ordinary_clone_previous_SELL[i]
```

Town/shop demand and an opponent BUY can only reduce this residual.
A positive remainder is therefore evidence that a near-clone
supplied more than its ordinary previous slot.

Promotion requires all of the following:

- public farm distance remains at most 2 after a 24-turn near-clone streak;
- the evidence appears only during steps 144–159, before v44 opens timing debt;
- the item is STRAWBERRY, MELON, MILK, or WOOL;
- the base route really has that item in the next two SELL slots;
- unexplained supply is **2–4 units**.

Only then is the game latched from horizon 2 to horizon 3. The
decision occurs before either controller has outstanding timing
debt, so switching cannot duplicate lifetime production.

## 8. Runtime invariants

- one complete v43 farm route is always active; no arbitrary suffix splice;
- visible first/second/selected third YARN events own the sparse farm branches;
- ordinary near-clones use the frozen v44 two-turn, ten-unit control;
- promoted near-clones use three turns and at most twenty units;
- every early SELL is borrowed from a real future route slot;
- the exact item/quantity debt is removed from that due slot;
- route changes clear outstanding debt instead of applying it to a different schedule;
- terminal liquidation is legal and collision-aware;
- no name, team, submission, score, episode, Notebook, or private opponent field enters runtime;
- no future opponent action or evaluation lineage label enters runtime;
- standalone `main.py` uses only the Python standard library.

## 9. Lineage split is evaluation metadata, not a runtime shortcut

Seed- or team-grouped folds leak badly when many teams replay one
public ancestor. The strict panel therefore identifies an opening
with actor h96 and defines continuation lineage from the actor
h96→h200 and h200→h360 suffixes. A holdout row must have a known
opening but an unseen composite continuation.

This follows the population-scale correction described by
[Georgy Mamarin](https://www.kaggle.com/datasets/georgymamarin/kaggriculture-episodes):
identical action streams can span many nominal teams. The lineage
label constructs folds only. The submitted policy never sees it.

## 12. Limits and the next test

The 60/64 result contains only four recent submissions. Frozen
action streams are exact for open-loop behavior but approximate for
adaptive opponents, and local engine/shop trajectories can diverge
after counterfactual actions. Publication may also create a new
phase that this detector has never observed.

The next experiment should not blindly move to horizon 4. It should
capture a new chronological block, estimate incremental value of
horizons 1/2/3 from public market flow, and require the same
no-regression checks against broad, seat, and continuation floors.
